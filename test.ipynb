```python
from pydantic import BaseModel
class UserClass(BaseModel):
    name: str
    email: str
    rollno: int
    is_active: bool
new_class = UserClass(name = 'vicky',email="vicky@gmail.com",rollno="1223",is_active=0)
print(new_class.model_dump())
```

```python
from typing import Optional
from pydantic import BaseModel,Field

class LMConfig(BaseModel):
    model_name:str = "gpt-4o"
    temperature:float = 0.7
    max_tokens : Optional[int] =None
    stop_sequences : list[str] = Field(default_factory=list)

config = LMConfig()
print(config.model_dump())
```

```python

from pydantic import BaseModel

class Message(BaseModel):
    role:str
    content :str

class ChatTurn(BaseModel):
    messages:list[Message]
    metadata: dict[str,str] = {}

turn  = ChatTurn(
    messages = [
        Message(role="user",content="Hello!"),
        Message(role="assistant",content="How can i help u today")
    ]
)
print(turn.model_dump_json(indent=2))
```

```python
from pydantic import BaseModel,  Field
class GenerationParams(BaseModel):
    temperature:float =Field(default=0.7,ge = 0.0,le=2.0,description="Sampling temperature")
    max_tokens: int = Field(default=512,gt=0)
    stop_sequences:list[str] = Field(default_factory=list)
try:
    params = GenerationParams(temperature=0.8,max_tokens=216)
    print(params.model_dump_json())
except Exception as e:
    print(e)
```

```python
from pydantic import BaseModel, field_validator

class PromptTemplate(BaseModel):
    template_text:str

    @field_validator("template_text")
    @classmethod
    def must_contain_variable(cls,v:str) -> str:
        if "+" not in v :
            raise ValueError("Template must contain an '+' placeholder.")
        return v
prompt = PromptTemplate(template_text="Hello+World")
        
```
